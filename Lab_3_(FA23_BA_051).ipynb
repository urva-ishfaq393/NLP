## Task 1: Regex Fundamentals (Character Classes & Quantifiers)

Given a sample text with mixed content (words, numbers, symbols), write regex patterns for:

**a)** Extract all digits from the text  
**b)** Extract all words that start with a capital letter  
**c)** Extract all sequences of exactly 4 digits (e.g., years)  
**d)** Extract all words that are between 3 to 6 characters long

```python
text1 = ("Ahmed moved to Lahore in 2019 and joined Google in 2023 with 150 colleagues. "
         "He paid $4500 for rent #12 & bought 3 laptops; Sara said, 'Great deal!' on 12345 items.")

# a) every single digit
print("a)", re.findall(r"\d", text1))

# b) words starting with a capital letter
print("b)", re.findall(r"\b[A-Z][a-zA-Z]*\b", text1))

# c) exactly 4 digits (not part of a longer number like 12345)
print("c)", re.findall(r"(?<!\d)\d{4}(?!\d)", text1))

# d) words that are 3 to 6 letters long
print("d)", re.findall(r"\b[A-Za-z]{3,6}\b", text1))
```

## Task 2: Pattern-Based Extraction (Real World Data)

Given a paragraph containing emails, phone numbers, order IDs, dates, and percentages mixed together:

**a)** Write a regex pattern using `re.findall()` to extract all email addresses  
**b)** Write a regex pattern to extract all phone numbers (assume format: `03XX-XXXXXXX`)  
**c)** Write a regex pattern to extract all dates in `DD/MM/YYYY` format  
**d)** Write a regex pattern to extract all percentage values (e.g., `45%`, `12.5%`)

```python
para2 = ("Order ORD-20481 was placed on 15/03/2024 by ali.khan@example.com, who can be reached at 0312-3456789. "
         "A second order, ORD-20499, came on 02/04/2024 from sara_99@mail.co.uk (phone 0345-1234567). "
         "Support: help@shop.pk. Sales rose by 45% in March and 12.5% in April, while returns fell to 3.75%. "
         "Ignore the invalid number 0412-3456789 and the date 2024/03/15.")

emails = re.findall(r"[\w.+-]+@[\w-]+(?:\.[\w-]+)+", para2)
phones = re.findall(r"\b03\d{2}-\d{7}\b", para2)
dates  = re.findall(r"\b\d{2}/\d{2}/\d{4}\b", para2)
percs  = re.findall(r"\b\d+(?:\.\d+)?%", para2)

print("a) Emails:     ", emails)
print("b) Phones:     ", phones)
print("c) Dates:      ", dates)
print("d) Percentages:", percs)
```

## Task 3: Word Tokenization (NLTK vs spaCy)

Using a given paragraph of your choice (at least 3-4 sentences, including contractions like "don't", "it's"):

**a)** Tokenize the paragraph using NLTK's `word_tokenize()`  
**b)** Tokenize the same paragraph using a blank spaCy pipeline  
**c)** Compare and write down (in a comment) at least 2 differences you noticed between the two outputs  
**d)** Count the total number of tokens produced by each method

```python
para3 = ("I don't think it's easy to build a state-of-the-art system. Dr. Khan said, \"We can't wait!\" "
         "Email info@site.com costs $50 for 2 users at 3.5% in the U.S.")

# a) NLTK
nltk_tokens = word_tokenize(para3)
print("NLTK :", nltk_tokens)

# b) blank spaCy pipeline
nlp_blank = spacy.blank("en")
spacy_tokens = [t.text for t in nlp_blank(para3)]
print("spaCy:", spacy_tokens)
```

```python
# c) Differences observed:
# 1. Hyphenated words: NLTK keeps 'state-of-the-art' as ONE token, spaCy splits it into
#    'state', '-', 'of', '-', 'the', '-', 'art'.
# 2. Emails: NLTK breaks 'info@site.com' into 'info', '@', 'site.com', while spaCy keeps it as ONE token.
# 3. Quotes: NLTK converts double quotes into `` and '' ; spaCy keeps the original " characters.
# 4. Abbreviations: NLTK gives 'U.S' + '.' (the final period is split off), spaCy keeps 'U.S.' whole.
# Both split contractions the same way: "don't" -> do + n't, "it's" -> it + 's.

# d) token counts
print("NLTK token count :", len(nltk_tokens))
print("spaCy token count:", len(spacy_tokens))
```

## Task 4: Token Attributes (Categorizing Tokens)

Using a sample sentence that contains normal words, numbers, an email address, and a currency symbol (e.g., "$50" or "Rs. 500"):

**a)** Load the sentence into a spaCy `Doc` object  
**b)** Loop through each token and print its `is_alpha` value  
**c)** Print each token's `like_num` and `like_email` values  
**d)** Identify and list only the tokens where `is_currency` is `True`

```python
# a)
sentence4 = "Mr. Ali paid $50 and Rs. 500 for 3 items; write to ali@example.com now."
doc4 = nlp_blank(sentence4)
print("Doc created with", len(doc4), "tokens\n")

# b)
print("is_alpha values:")
for token in doc4:
    print(f"  {token.text:<18}{token.is_alpha}")

# c)
print("\nlike_num / like_email values:")
print(f"  {'Token':<18}{'like_num':<10}like_email")
for token in doc4:
    print(f"  {token.text:<18}{str(token.like_num):<10}{token.like_email}")

# d)
currency_tokens = [token.text for token in doc4 if token.is_currency]
print("\nCurrency tokens:", currency_tokens)
# Note: spaCy flags the symbol '$' but NOT the abbreviation 'Rs', because is_currency only recognises currency symbols.
```

## Task 5: Sentence Segmentation (Handling Tricky Cases)

Using a paragraph that includes at least one abbreviation (e.g., "Dr.", "Mr.", "e.g.") in the middle of a sentence:

**a)** Load a trained spaCy pipeline (`en_core_web_sm`)  
**b)** Use `doc.sents` to split the paragraph into individual sentences  
**c)** Print the total number of sentences detected  
**d)** Check whether the abbreviation caused any incorrect sentence break, and explain your observation in one line

```python
# a)
nlp_sm = spacy.load("en_core_web_sm")

para5 = ("Dr. Smith works at the city clinic. He treats many patients, e.g. children and adults. "
         "Mr. Jones visited him on Monday. Prof. Lee called later.")
doc5 = nlp_sm(para5)

# b)
sentences = [s.text.strip() for s in doc5.sents]
for i, s in enumerate(sentences, 1):
    print(f"{i}. {s}")

# c)
print("\nTotal sentences detected:", len(sentences))

# d) a wrong break would leave a sentence that is just an abbreviation (e.g. 'Dr.') or ends with one
abbrevs = {"Dr.", "Mr.", "Mrs.", "Ms.", "Prof.", "e.g.", "i.e."}
bad = [s for s in sentences if s.split()[-1] in abbrevs]
expected = 4
if not bad and len(sentences) == expected:
    print("Observation: No incorrect break - spaCy's tokenizer and parser treat 'Dr.', 'Mr.', 'Prof.' and 'e.g.' as part of the sentence, not as sentence endings.")
else:
    print("Observation: An abbreviation caused a wrong break in:", bad)
```
